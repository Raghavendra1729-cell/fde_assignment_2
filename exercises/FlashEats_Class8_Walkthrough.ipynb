# 2. Turn the notebook into stages

We will use five explicit stages:

```text
EXTRACT → VALIDATE → CLEAN → TRANSFORM → SAVE
```

Why stages?

Because when something fails, we want to know **what responsibility failed**.

Open the actual project structure:

````python
for path in sorted(PROJECT.rglob("*")):
    if path.is_file():
        print(path.relative_to(PROJECT))
````

### Discussion

- Why should validation happen before transformation?
- Why preserve raw API responses?
- Why is `save()` a separate stage?
- Would you put every operation in one giant function?

### Deduction

> Functions are not the goal. **Isolation of responsibility** is the goal.

**My answers**

- **Validation before transformation:** if `promised_eta` is missing, `build_order_journey` would either crash halfway or fill `delay_min` with NaN and still give out a late rate. Checking first stops the run before anything is built from bad input.
- **Preserve raw API responses:** the dispatch API only gives the current state. If a number looks wrong next week we can check exactly what dispatch sent that day without calling it again. Same idea as the raw pages I saved in Class 5.
- **save() as its own stage:** it is the only step that changes what other people read, so it should only run after everything else passed. The atomic write logic also sits in one place.
- **One giant function?** No. When it fails you get one long traceback and have to read everything. With stages the log already says which step stopped, and each piece can be tested on its own.

### Discussion

1. Which values could differ between local, staging and production?
2. Which values should never be committed if they are secrets?
3. Why is `MAX_RETRIES` configuration but retry logic code?

### Deduction

> A pipeline is easier to move between environments when deployment-specific values are outside core business logic.

**My answers**

1. `DISPATCH_API_URL` (localhost here, the real dispatch host in production), `START_MOCK_API` (true only on my laptop, production should never start a mock), `LOG_LEVEL`, `PAGE_SIZE`, `MAX_DATA_AGE_DAYS` and `MAX_RETRIES` could all be different per environment.
2. Nothing in `.env.example` is secret right now. A real dispatch API would need a token and the database would need a password. Those go in a `.env` that is not committed (or a secrets manager). Only the example file with dummy values gets committed.
3. How many times to retry is a tuning number that can change per environment without a code change. How to retry (which status codes, the backoff, when to give up) is behaviour, so it should be reviewed and tested like the rest of the code.

# 5. Retries: retry the right failures

The FlashEats Dispatch API deliberately behaves like a real external dependency:

- page 3 fails once with HTTP 500,
- page 5 rate-limits once with HTTP 429,
- pagination must be completed.

A retry is appropriate for a **transient** failure.

A retry is usually not useful for:

- missing required column,
- invalid credentials,
- incompatible schema,
- a business-rule failure.

Let's inspect the retry logic.

````python
extract_code = (PROJECT / "pipeline" / "extract.py").read_text()

start = extract_code.find("def _request_page")
end = extract_code.find("def extract_dispatch_api")

print(extract_code[start:end])
````

### Discussion

1. Why limit the number of retries?
2. Why wait longer between repeated attempts?
3. Why should HTTP 429 and HTTP 500 be treated differently from a schema failure?
4. Could retrying a write operation be dangerous?

### Deduction

> **Retry transient failures; fail fast on conditions that require human or code changes.**

**My answers**

1. If dispatch is really down, unlimited retries would hang the job and nobody would notice. It also keeps hitting an API that is already rate limiting us. With `MAX_RETRIES=3` the job gives up after a few seconds with a clear error.
2. Waiting longer each time (`base_seconds * 2 ** (attempt - 1)`, so 1s, 2s, 4s) gives the other system time to recover instead of flooding it. For 429 the code uses the server's `retry_after_seconds`, which is even better because the server tells us how long.
3. 500 and 429 are about the other system at that moment, so a second try can work. A missing column will still be missing on the next try, so retrying only wastes time and hides the real error. Other 4xx codes raise `NonRetryableAPIError` straight away.
4. Yes. Our calls are GETs, so repeating them is safe. Retrying something like "insert rows" or "issue refund" after a timeout could do it twice if the first call actually went through. That is one reason the save stage replaces files instead of appending.

# 6. Idempotency: what happens if we run twice?

A job that appends the same 1,600 orders every time is repeatable — but not dependable.

For this class:

```text
same logical run_date
        ↓
same output partition
        ↓
atomically replace output
```

Let's inspect the save stage.

````python
save_code = (PROJECT / "pipeline" / "save.py").read_text()
print(save_code)
````

### Discussion

1. Why partition outputs by logical run date?
2. Why replace instead of append for this pipeline?
3. When would append actually be correct?
4. Why write a temporary file before replacing the final file?

### Deduction

> **A rerun should not create unintended duplicate effects.**

**My answers**

1. Each logical date gets its own folder, so rerunning 2026-09-22 only touches that folder and other days stay as they were. Backfilling an old day is just a rerun with that date.
2. Every run rebuilds the full order journey (all 1600 orders), it is a snapshot and not new events. Appending would give 3200 rows after a rerun and every metric would be double counted.
3. Append is correct when each run brings only new, non-overlapping records, for example an event log where every event has its own id and we only fetch events since the last run. Even then I would keep a dedup key.
4. If the job crashes in the middle of `to_csv`, the final file would be half written and someone could read it. Writing to a temp file in the same folder and then `os.replace` swaps it in one step, so a reader sees either the old file or the new one. One thing I noticed: only the CSV is written this way, `metrics.json` and `validation_report.json` use plain `write_text`.

# 9. Prove idempotency

Run exactly the same logical job again.

What should happen?

- output row count should not double,
- the same partition should be replaced,
- the final business dataset should remain one row per order.

````python
output_path = (
    PROJECT
    / "data"
    / "processed"
    / "run_date=2026-09-22"
    / "order_journey.csv"
)

first = pd.read_csv(output_path)
first_rows = len(first)

second_run = subprocess.run(
    [
        sys.executable,
        str(PROJECT / "run_pipeline.py"),
        "--run-date",
        "2026-09-22",
    ],
    cwd=PROJECT,
    text=True,
    capture_output=True,
)

second = pd.read_csv(output_path)
second_rows = len(second)

print("First run rows:", first_rows)
print("Second run rows:", second_rows)
print("Rows doubled?", second_rows == first_rows * 2)
print("Same logical row count?", first_rows == second_rows)
````

````python
# my extra check after the rerun
print("second run return code:", second_run.returncode)
print("unique order_id in output:", second["order_id"].nunique(), "of", len(second))
print("first and second output identical:", first.equals(second))
print("files in the partition:", sorted(p.name for p in output_path.parent.iterdir()))

raw_dir = PROJECT / "data" / "raw" / "dispatch" / "run_date=2026-09-22"
print("raw dispatch pages:", len(list(raw_dir.glob("*.json"))))
print(json.loads((output_path.parent / "metrics.json").read_text()))
````

Both runs gave the same row count and the rerun did not double anything. On top of the row count I checked that `order_id` is still unique, that the two outputs are exactly the same, and that the folder still has only the 3 expected files (no leftover temp file, no second copy). The raw folder still has 8 pages because the rerun overwrote the same page files.

### Deduction

> **“Can rerun” and “safe to rerun” are different properties.**

### Extra: the other chaos modes

`run_pipeline.py` also accepts `--chaos duplicate_order` and `--chaos stale_data`. I ran both, then one normal run at the end so the saved partition comes from a clean run again.

````python
def run_job(*extra):
    return subprocess.run(
        [sys.executable, str(PROJECT / "run_pipeline.py"), "--run-date", "2026-09-22", *extra],
        cwd=PROJECT,
        text=True,
        capture_output=True,
    )

keywords = ["CHAOS", "uniqueness", "freshness", "Cleaned orders", "stopped", "completed"]

for mode in ["duplicate_order", "stale_data"]:
    r = run_job("--chaos", mode)
    print(f"--- {mode} | return code {r.returncode}")
    print([line for line in r.stdout.splitlines() if "PIPELINE" in line])
    for line in r.stderr.splitlines():
        if any(k in line for k in keywords):
            print("   ", line.split(" | ", 1)[1][:200])
    print()
````

````python
final = run_job()
print("clean run return code:", final.returncode)
print("rows in order_journey.csv:", len(pd.read_csv(output_path)))
for check in json.loads((output_path.parent / "validation_report.json").read_text()):
    print(check["status"], "|", check["check"], "|", check["detail"])
````

- **duplicate_order:** return code 0 with a WARN. Duplicate rows involved went from 6 to 8, clean removed 4 rows instead of 3, and the output still has 1600 rows. So an extra duplicate is handled by the agreed one-row-per-order rule and does not stop the run. I think there should still be a limit, because a WARN is too soft if hundreds of duplicates suddenly appear.
- **stale_data:** return code 2, stopped at the freshness check (`age_days=390, allowed=60`) and nothing was written. It still called the API before failing, because validation only runs after all the extraction is done.
- After the last clean run the validation report is back to the normal values (6 duplicate rows, 37 delivered orders with no delivery time, data 25 days old).
